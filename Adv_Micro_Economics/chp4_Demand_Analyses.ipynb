```python

# HW3 - P4.8: Cross-Price Elasticity

# Given data
app_price_old = 12
app_price_new = 10

app_orders_old = 90
app_orders_new = 150

bev_sales_old = 300
bev_sales_new = 600

# -----------------------------
# Part A: Own-price elasticity
# -----------------------------

pct_change_app_quantity = (
    (app_orders_new - app_orders_old) / app_orders_old
) * 100

pct_change_app_price = (
    (app_price_new - app_price_old) / app_price_old
) * 100

price_elasticity = (
    pct_change_app_quantity / pct_change_app_price
)

print("PART A: Price Elasticity")
print(f"Appetizer quantity change: {pct_change_app_quantity:.2f}%")
print(f"Appetizer price change: {pct_change_app_price:.2f}%")
print(f"Price elasticity: {price_elasticity:.2f}\n")


# -----------------------------
# Part B: Cross-price elasticity
# -----------------------------

pct_change_bev_quantity = (
    (bev_sales_new - bev_sales_old) / bev_sales_old
) * 100

cross_price_elasticity = (
    pct_change_bev_quantity / pct_change_app_price
)

print("PART B: Cross-Price Elasticity")
print(f"Beverage quantity change: {pct_change_bev_quantity:.2f}%")
print(f"Cross-price elasticity: {cross_price_elasticity:.2f}")
print("Interpretation: Negative -> complementary goods.\n")


# -----------------------------
# Part C: Further reduction to $8
# -----------------------------

app_price_future = 8

pct_change_future_price = (
    (app_price_future - app_price_new) / app_price_new
) * 100

# Estimate future appetizer orders
pct_change_future_app_quantity = (
    price_elasticity * pct_change_future_price
)

app_orders_future = app_orders_new * (
    1 + pct_change_future_app_quantity / 100
)

# Estimate future beverage sales
pct_change_future_bev_quantity = (
    cross_price_elasticity * pct_change_future_price
)

bev_sales_future = bev_sales_new * (
    1 + pct_change_future_bev_quantity / 100
)

# Appetizer revenue
app_revenue_before = app_price_old * app_orders_old
app_revenue_after_first_cut = app_price_new * app_orders_new
app_revenue_future = app_price_future * app_orders_future

print("PART C: Revenue Analysis")
print(f"Appetizer revenue before: ${app_revenue_before:.2f}/day")
print(f"Appetizer revenue at $10: ${app_revenue_after_first_cut:.2f}/day")
print(f"Estimated orders at $8: {app_orders_future:.0f}")
print(f"Estimated appetizer revenue at $8: ${app_revenue_future:.2f}/day")
print(f"Estimated beverage sales: {bev_sales_future:.0f} units/day")
print("Beverage revenue rises if its price remains unchanged.")
```
