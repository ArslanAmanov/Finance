### Problem 1 SOC

Example 1: A firm's profit function is 
𝜋(𝑄)=−𝑄3 +9𝑄2 +48𝑄−100 
 
where Q is the firm's output. 
- (a) Find the critical point(s) of the profit function. 
- (b) Use the second-order condition (SOC) to determine whether the relevant critical point is a 
maximum or minimum. 
- (c) Calculate the firm's maximum profit. 

```python
# Profit function and derivatives
Q = sp.symbols('Q', real=True)
profit = -Q**3 + 9*Q**2 + 48*Q - 100

first_derivative = sp.diff(profit, Q)
second_derivative = sp.diff(first_derivative, Q)
critical_points = sp.solve(first_derivative, Q)

print("(a) Critical points:")
for point in critical_points:
    print(f"  Q = {point}")

print("\n(b) Second-order condition:")
for point in critical_points:
    print(f"  π''({point}) = {second_derivative.subs(Q, point)}")

print("\n(c) Maximum profit for nonnegative output:")
max_output = next(point for point in critical_points if point >= 0)
max_profit = profit.subs(Q, max_output)
print(f"  Output: Q = {max_output}")
print(f"  Maximum profit: π({max_output}) = {max_profit}")
```
