```python
# Supply functions

def olympia_supply(P):
    if P < 350:
        return 0
    return -7_000_000 + 20_000 * P


def yakima_supply(P):
    if P < 150:
        return 0
    return -750_000 + 5_000 * P


def industry_supply(P):
    return olympia_supply(P) + yakima_supply(P)


prices = [325, 350, 375]

for P in prices:
    Q_O = olympia_supply(P)
    Q_Y = yakima_supply(P)
    Q_industry = industry_supply(P)

    print(f"Price = ${P}")
    print(f"  Olympia = {Q_O:,.0f}")
    print(f"  Yakima  = {Q_Y:,.0f}")
    print(f"  Industry = {Q_industry:,.0f}")
    print()
```

```python
import numpy as np
import matplotlib.pyplot as plt

# Price range
P = np.linspace(150, 500, 500)

# Firm supply
Q_yakima = np.maximum(0, -750_000 + 5_000 * P)
Q_olympia = np.maximum(0, -7_000_000 + 20_000 * P)

# Industry supply
Q_industry = Q_yakima + Q_olympia

plt.figure(figsize=(10, 6))

plt.plot(Q_yakima, P, label="Yakima")
plt.plot(Q_olympia, P, label="Olympia")
plt.plot(Q_industry, P, label="Industry")

plt.xlabel("Quantity (1,000 board feet)")
plt.ylabel("Price ($)")
plt.title("Supply Curves: Olympia, Yakima, and Industry")

plt.axhline(350, linestyle="--", alpha=0.5)
plt.axvline(1_000_000, linestyle="--", alpha=0.5)

plt.legend()
plt.grid(True)
plt.show()
```

### Eye-de-ho Potato Example

```python
# Given values
PW = 4
Y = 15_000
PL = 8
R = 20

# Demand function
def demand(P):
    return -1450 - 25*P + 12.5*PW + 0.1*Y

# Supply function
def supply(P):
    return -100 + 75*P - 25*PW - 12.5*PL + 10*R


# Part A
print("Demand function:")
print("QD = 100 - 25P")

print("\nSupply function:")
print("QS = -100 + 75P")
```

```python
prices = [1.50, 2.00, 2.50]

for P in prices:

    QD = demand(P)
    QS = supply(P)

    difference = QS - QD

    print(f"\nPrice = ${P:.2f}")
    print(f"Quantity Demanded = {QD:.2f} million")
    print(f"Quantity Supplied = {QS:.2f} million")

    if difference > 0:
        print(f"Surplus = {difference:.2f} million")
    elif difference < 0:
        print(f"Shortage = {-difference:.2f} million")
    else:
        print("Equilibrium: No surplus or shortage")
```

```python
import sympy as sp

P = sp.symbols('P')

QD = 100 - 25*P
QS = -100 + 75*P

P_star = sp.solve(QD - QS, P)[0]

Q_star = QD.subs(P, P_star)

print("Equilibrium price: $",P_star)
print("Equilibrium quantity:", Q_star, "million")
```
