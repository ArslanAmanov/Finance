```python
# create me dummy dataset or library datset to use for inferential statistics 

import pandas as pd
import numpy as np

df = pd.DataFrame({
    'A': np.random.normal(loc=0, scale=1, size=100),
    'B': np.random.normal(loc=5, scale=2, size=100)
})        
```

```python
print(df.head())
```

```python
#  t-test for independent samples 
from scipy import stats
t_stat, p_value = stats.ttest_ind(df['A'], df['B'])
print("t-statistic:", t_stat)
print("p-value:", p_value) 
```

```python
# plot the distributions of the two samples
import matplotlib.pyplot as plt
plt.hist(df['A'], bins=20, alpha=0.5, label='A')
plt.hist(df['B'], bins=20, alpha=0.5, label='B')
plt.legend()
plt.title('Distributions of Samples A and B')
plt.xlabel('Value')
plt.ylabel('Frequency')
plt.show()
```
