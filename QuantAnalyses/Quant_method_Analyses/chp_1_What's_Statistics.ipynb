# What is Statistics? 

Statistics is a branch of mathematics that deals with the collection, analysis, interpretation, presentation, and organization of data. It provides tools and methods for making sense of complex data sets, allowing researchers and analysts to draw meaningful conclusions and make informed decisions based on empirical evidence. 

Examples of statistics practitioners include the following:
1. A financial analyst who uses statistical models to predict stock market trends and make investment decisions. 
2. An economist who analyzes economic data to forecast inflation rates and unemployment trends. 
3. A market researcher who conducts surveys and analyzes consumer behavior to inform product development and marketing strategies. 

# This discipline can be broadly categorized into two main branches: 

- **Descriptive statistics:** This branch focuses on summarizing and describing the main features of a data set, including measures of central tendency, variability, and graphical representations.


- **Inferential statistics:** This branch involves making predictions, generalizations, and inferences about a population based on a sample of data drawn from that population.

Statistical inference problems involve three key concepts: 
1. **Population:** The entire group of individuals or items that we want to study or make conclusions about. For example, if we are interested in the average income of all adults in a country, the population would be all adults in that country. A descriptive measure of the population is called **a parameter**.

2. **Sample:** A subset of the population that is selected for analysis. The sample should be representative of the population to ensure accurate inferences. For example, we might randomly select 1,000 adults from the country to estimate the average income.

3. **Parameter:** A numerical characteristic of the population, such as the population mean or population standard deviation. Parameters are typically unknown and are estimated using sample statistics.    

Statistical inference is a measure of reliability. There are two such measures: **the confidence level and the significance level.**
The confidence level is the probability that a population parameter lies within a specified range of values, while the significance level is the probability of rejecting a null hypothesis when it is actually true.

For example, if we conduct a hypothesis test with a significance level of 0.05, we are willing to accept a 5% chance of incorrectly rejecting the null hypothesis when it is true. Conversely, if we construct a confidence interval with a confidence level of 95%, we are stating that we are 95% confident that the true population parameter lies within the specified range of values.

A politician who is running for the office of mayor of a city with 25,000 registered voters comissions a survey. 
In the survey 48% of the 200 registered voters interviewed say they plan to vote for the politician. 

a. What is the population of interest ?

**Answer: The population of interest is all 25,000 registered voters in the city.**

b. What is the sample ? 

**Answer: The sample is the 200 registered voters who were interviewed in the survey.**

c. Is the value 48% a parameter or a statistic ? 

**Answer: The value 48% is a statistic because it is a measure calculated from the sample data, not the entire population.**

**Exercise 1.4** A manufacturer of computer chips claims that less than 10% of its products are defective. When 1,000 chips were drawn from a large production, 7.5% were found to be defective. 

a. What is the population of interest ?

**Answer: The population of interest is all the computer chips produced by the manufacturer.** 

b. What is the sample ?

**Answer: The sample is the 1,000 computer chips that were drawn from the production for testing.** 

c. What is the parameter? 

**Answer: The parameter is the true proportion of defective computer chips in the entire production, which is claimed to be less than 10%.**

d. What is the statistic?

**Answer: The statistic is the proportion of defective computer chips found in the sample, which is 7.5%.** 

e. Does the value 10% represent a parameter or a statistic?

**Answer: The value 10% represents a parameter because it is a claim about the entire population of computer chips produced by the manufacturer.**  

f. Is the value 7.5% a parameter or a statistic? 

**Answer: The value 7.5% is a statistic because it is a measure calculated from the sample data, not the entire population.** 

g. Explain briefly how the statistic can be used to make inferences about the parameter to test the claim. 

**Answer: The statistic (7.5% defective chips in the sample) can be compared to the claimed parameter (less than 10% defective chips in the population). If the sample statistic is significantly lower than the claimed parameter, it provides evidence to support the manufacturer's claim. Statistical tests, such as hypothesis testing, can be conducted to determine if the observed sample statistic is consistent with the claimed parameter or if it suggests that the claim may not be accurate.**

Suppose you believe that, in general, graduates who have majored in your subject are offered higher salaries upon graduating than are graduates of other programs. Describe a statistical experiment that could help test your belief. 

**Answer: To test the belief that graduates who have majored in my subject are offered higher salaries than graduates of other programs, I could design a statistical experiment as follows:** 

1. **Define the Population:** Identify the population of interest, which includes all graduates from my subject and graduates from other programs.

2. **Select a Sample:** Randomly select a representative sample of graduates from my subject and a comparable sample of graduates from other programs. We will ensure that the sample sizes are sufficient to provide reliable results.

3. **Collect Data:** Gather data on the starting salaries of the graduates in both samples. We could do this through surveys, alumni records, or employment databases. 

4. **Analyze the Data:** We calculate the average starting salary for graduates from my subject and the average starting salary for graduates from other programs. We will need to use statistical tests, such as a t-test, to determine if there is a significant difference between the two averages. 

5. **Draw Conclusions:** Based on the results of the statistical analysis, determine whether there is sufficient evidence to support the belief that graduates from my subject are offered higher salaries than graduates from other programs. If the difference is statistically significant, it would suggest that my belief is valid; if not, it may indicate that there is no substantial difference in starting salaries between the two groups.
