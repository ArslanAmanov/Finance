# Random Experiment:
is an action or process that leads to one of several possible outcomes.

- The first step in assigning probabilities is to produce a list of the outcomes. 
- The listed outcomes must be exhaustive, which means that all possible outcomes must be included. 
- In addition, the outcomes must be mutually exclusive, which means that no 
two outcomes can occur at the same time.

# Sample Space 
A list of exhaustive and mutually exclusive outcomes is called a sample space. The sample space is usually denoted by the symbol S.

# Three Approaches to Assigning Probabilities

- 1. **Classical Approach:** The probability of an event is determined by the ratio of the number 
of favorable outcomes to the total number of possible outcomes, assuming all outcomes are equally likely.

- 2. **Relative Frequency Approach:** The probability of an event is determined by the 
ratio of the number of times the event occurs to the total number of trials.
- 3. **Subjective Approach:** The probability of an event is based on personal judgment, intuition, or experience rather than on exact calculations.
