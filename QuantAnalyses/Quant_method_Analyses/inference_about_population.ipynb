```python
import numpy as np
from scipy import stats

data = np.array([16, 23, 18, 41, 22, 18, 23, 19, 22, 15, 
                18, 35, 16, 15, 17, 19, 23, 15, 16, 26])

n = len(data)
x_bar = np.mean(data)
s = np.std(data, ddof=1)
mu_0 = 20
alpha = 0.05

# One-sample t-test (right-tailed)
t_stat, p_val_two_sided = stats.ttest_1samp(data, mu_0, alternative='greater')
t_crit = stats.t.ppf(1 - alpha, df=n - 1)

print(f"Sample Mean (x_bar) = {x_bar:.2f}")
print(f"Sample Std Dev (s)  = {s:.4f}")
print(f"t-statistic         = {t_stat:.4f}")
print(f"Critical value      = {t_crit:.4f}")
print(f"p-value             = {t_stat if False else p_val_two_sided:.4f}")

if p_val_two_sided < alpha:
    print("Decision: Reject H0 — Sufficient evidence supporting the doctor's claim.")
else:
    print("Decision: Fail to reject H0 — Insufficient evidence to support the doctor's claim.")
```

```python
from scipy import stats

def chi2_two_tailed_test(n, s2, sigma2_0, alpha=0.05):
    df = n - 1
    chi2_stat = (df * s2) / sigma2_0
    
    # Two-tailed critical values
    crit_lower = stats.chi2.ppf(alpha / 2, df)
    crit_upper = stats.chi2.ppf(1 - alpha / 2, df)
    
    # Two-tailed p-value
    p_lower = stats.chi2.cdf(chi2_stat, df)
    p_upper = 1 - stats.chi2.cdf(chi2_stat, df)
    p_val = 2 * min(p_lower, p_upper)
    
    print(f"--- Sample Size n = {n} (df = {df}) ---")
    print(f"Chi-square Statistic : {chi2_stat:.4f}")
    print(f"Critical Values       : [{crit_lower:.4f}, {crit_upper:.4f}]")
    print(f"p-value              : {p_val:.4f}")
    decision = "Reject H0" if (chi2_stat < crit_lower or chi2_stat > crit_upper) else "Fail to reject H0"
    print(f"Decision             : {decision}\n")

# Part (a)
chi2_two_tailed_test(n=100, s2=220, sigma2_0=300, alpha=0.05)

# Part (b)
chi2_two_tailed_test(n=50, s2=220, sigma2_0=300, alpha=0.05)
```

```python
import numpy as np
from scipy import stats

# Given values
n = 200
x = 41
p0 = 0.25
alpha = 0.10

# Sample proportion
p_hat = x / n

# Standard error under H0
se = np.sqrt(p0 * (1 - p0) / n)

# z-statistic
z_stat = (p_hat - p0) / se

# Critical value and p-value (left-tailed test)
z_crit = stats.norm.ppf(alpha)
p_val = stats.norm.cdf(z_stat)

print(f"Sample Proportion (p_hat) : {p_hat:.4f}")
print(f"Standard Error (SE)       : {se:.5f}")
print(f"z-statistic               : {z_stat:.4f}")
print(f"Critical Value (z_crit)   : {z_crit:.4f}")
print(f"p-value                   : {p_val:.4f}")

if z_stat < z_crit:
    print("\nDecision: Reject H0 — Sufficient evidence at 10% level to support the officer's belief.")
else:
    print("\nDecision: Fail to reject H0 — Insufficient evidence to support the officer's belief.")
```
