```python
import numpy as np 
import scipy.stats as stats


# Normal distribution
mean = 175
std = 8

# PDF at x = 180
pdf_value = stats.norm.pdf(180, loc=mean, scale=std)

print("PDF:", pdf_value)

# P(X <= 180)
cdf_value = stats.norm.cdf(180, loc=mean, scale=std)

print("CDF:", cdf_value)

# P(170 <= X <= 180)
probability = (stats.norm.cdf(180, loc=mean, scale=std) - stats.norm.cdf(170, loc=mean, scale=std))

print("Probability:", probability)
```

```python
import numpy as np
import scipy.stats as stats

np.random.seed(42)

heights = stats.norm.rvs(
    loc=175,
    scale=8,
    size=1000
)

print(heights[:10])
```

```python
print("Mean:", np.mean(heights)) 
print("Standard Deviation:", np.std(heights)) 
print("Minimum:", np.min(heights)) 
print("Maximum:", np.max(heights)) 
```

```python
import numpy as np
import matplotlib.pyplot as plt
import scipy.stats as stats

x = np.linspace(140, 210, 500)

pdf = stats.norm.pdf(
    x,
    loc=175,
    scale=8
)

plt.plot(x, pdf)

plt.xlabel("Height (cm)")
plt.ylabel("Probability Density")
plt.title("Normal Distribution of Height")

plt.show()
```

```python
import scipy.stats as stats

mean_return = 0.001      # 0.1%
std_return = 0.02        # 2%

# Probability of losing more than 5%
prob_loss = stats.norm.cdf(
    -0.05,
    loc=mean_return,
    scale=std_return
)

print(prob_loss) 
```

```python
import scipy.stats as stats

mean = 10
std = 0.1

probability = (
    stats.norm.cdf(10.2, mean, std)
    - stats.norm.cdf(9.8, mean, std)
)

print(probability)
```

```python
import yfinance as yf
import scipy.stats as stats

data = yf.download("AAPL", period="1y")

returns = data["Close"].pct_change().dropna()

mu, sigma = stats.norm.fit(returns)

print("Mean:", mu)
print("Std:", sigma)
```

```python
returns_95 = stats.norm.ppf(0.05, mu, sigma)
```

```python
print ("5% quantile of returns:", returns_95) 
```

```python
probability = stats.norm.cdf(
    -0.05,
    loc=mu,
    scale=sigma
)

print("P(Return < -5%):", probability)
```
