### Chapter 14: ANOVA (Analysis of Variance)

````python
#Two way ANOVA 

import pandas as pd
from statsmodels.formula.api import ols
from statsmodels.stats.anova import anova_lm
# Example data for two-way ANOVA

df= pd.DataFrame({
    'score': [
        70,72,75,78,80,82,
        65,68,70,71,73,75],
    'method': [
        'Traditional','Traditional','Online',
        'Online','Blended','Blended',
        'Traditional','Traditional','Online',
        'Online','Blended','Blended'],
    'class_size': [
        'Small','Small','Small',
        'Small','Small','Small',
        'Large','Large','Large',
        'Large','Large','Large'
    ]
})
````

````python
import statsmodels.api as sm
from statsmodels.formula.api import ols

model = ols(
    'score ~ C(method) * C(class_size)',
    data=df
).fit()

anova_table = sm.stats.anova_lm(model, typ=2)

print(anova_table)
````

Question 1 — One-way ANOVA

A company tests three sales training programs.


````python
# Determine:
# Whether the average sales performance differs between training programs.
# The ANOVA F-statistic.
# The p-value.
# If significant, perform Tukey HSD.
# Identify which training programs differ.


import pandas as pd
import statsmodels.api as sm
from statsmodels.formula.api import ols
from statsmodels.stats.multicomp import pairwise_tukeyhsd


df = pd.DataFrame({
    'score': [52, 55, 53, 54, 50, 56, 51, 60, 62, 61, 63, 59, 64, 62, 55, 57, 56, 58, 54, 59, 57],
    'program': ['A']*7 + ['B']*7 + ['C']*7
})

# Perform one-way ANOVA
model = ols('score ~ C(program)', data=df).fit()
anova_table = sm.stats.anova_lm(model, typ=2)
print(anova_table)

# If significant, perform Tukey HSD
tukey = pairwise_tukeyhsd(endog=df['score'], groups=df['program'], alpha=0.05)
print(tukey)

# Identify which training programs differ
print(tukey.summary())
````

Interpretation: 
Based on the ANOVA results, if the p-value is less than 0.05, we reject the null hypothesis and conclude that there are significant differences in average sales performance between the training programs. If the p-value is greater than 0.05, we fail to reject the null hypothesis and conclude that there is no significant difference in average sales performance between the training programs. 

If the ANOVA is significant, the Tukey HSD test helps identify which specific training programs differ from each other. The summary of the Tukey HSD test will show pairwise comparisons, the mean differences, and whether the differences are statistically significant. 

### Chi-Square Test: 
The Chi-Square test is used to determine whether there is a significant association between two categorical variables. It compares the observed frequencies in each category to the frequencies we would expect if there were no association between the variables.

In practice, you would create a contingency table of the two categorical variables and then apply the Chi-Square test to see if the observed distribution differs significantly from the expected distribution under the null hypothesis of no association.

````python
# create contingency table for Chi-Square test 

contingency_table = pd.crosstab(df['program'], df['score']) 
print(contingency_table)
````

````python
### Chi-Square Test Implementation
import pandas as pd
from scipy.stats import chi2_contingency

# perform Chi-Square test
chi2, p, dof, expected = chi2_contingency(contingency_table)
print(f"Chi2: {chi2}, p-value: {p}, Degrees of Freedom: {dof}")
print("Expected Frequencies:")
print(expected)
````

````python
from scipy.stats import chisquare

observed = [25, 35, 20, 20]
expected = [25, 25, 25, 25]

chi2, p = chisquare(
    f_obs=observed,
    f_exp=expected
)

print("Chi-square:", chi2)
print("p-value:", p)
````

## The key decision framework

This is probably the most useful way to memorize these for your QMEA exam:

### 1. One numerical outcome + independent groups → One-Way ANOVA

Ask:

> **Am I comparing means?**

Example:

> Productivity score of employees in Training A, B, C, D.

```text
Numerical outcome
       ↓
Compare means
       ↓
Different/independent people
       ↓
ONE-WAY ANOVA
```

**Example:**
50 patients at Hospital A, 50 different patients at Hospital B, 50 different patients at Hospital C.

→ **One-Way ANOVA**

---

### 2. Numerical outcome + same subjects/related groups → Randomized Block ANOVA

The key question:

> **Does the same person/object appear in multiple treatment groups?**

Example:

> 20 students try Method A, B, and C.

The students naturally differ in ability.

```text
Students
   ↓
Each student experiences
A + B + C
   ↓
Student = BLOCK
Method = TREATMENT
   ↓
Randomized Block ANOVA
```

This is essentially controlling for the variation caused by the **student**.

### Easy memory trick

**Same person repeatedly → BLOCK**

Examples:

* Same students → block
* Same patients → block
* Same technicians → block
* Same machines → potentially block, depending on the research question

---

### 3. One categorical variable + expected proportions → Goodness-of-Fit

Ask:

> **Do my observed categories match some expected percentages?**

Example:

A company claims:

```text
Package A = 10%
Package B = 20%
Package C = 30%
Package D = 20%
Package E = 20%
```

You collect actual customer choices.

You are testing:

$$
Observed \quad vs. \quad Expected
$$

→ **Chi-Square Goodness-of-Fit**

### Keyword

**Expected proportions → Goodness-of-Fit**

---

### 4. Two categorical variables → Chi-Square Independence

Ask:

> **Do I have two categorical variables and want to know whether they are associated?**

Example:

|          | Cash | Card | QR |
| -------- | ---: | ---: | -: |
| Under 30 |      |      |    |
| 30–49    |      |      |    |
| 50+      |      |      |    |

Variables:

* Age group = categorical
* Payment method = categorical

Question:

> Is payment method associated with age group?

→ **Chi-Square Test of Independence**

# Q2 is particularly important

Suppose you have:

$$
H_0:\mu_1=\mu_2=\mu_3=\mu_4
$$

ANOVA gives:

$$
p < 0.05
$$

You reject \(H_0\).

The correct conclusion is:

> **At least one population mean differs from another.**

You **cannot** immediately say:

> "All four means are different."

You need a **post-hoc test**, such as **Tukey's HSD**, to determine which pairs differ.

For example:

```text
ANOVA significant
       ↓
At least one mean differs
       ↓
Tukey HSD
       ↓
Determine which pairs differ
```

````python
# One -way ANOVA 

import pandas as pd
from scipy import stats

# Example data  
data = {
    'Finance': [3, 1, 4, 1],
    'Marketing': [1, 5, 3, 4],
    'Management': [8, 5, 4, 6]
}

df = pd.DataFrame(data)

# Perform one-way ANOVA 
# Ho: All group means are equal (null hypothesis)
# Ha: At least one group mean is different (alternative hypothesis)
f_statistic, p_value = stats.f_oneway(df['Finance'], df['Marketing'], df['Management'])
print("F-statistic:", f_statistic)
print("P-value:", p_value) 
print("Reject null hypothesis" if p_value < 0.05 else "Fail to reject null hypothesis")
````

Conclusion  At least 5% significance level, the data provide sufficient evidence that the average number of job offers differs by MBA major. Management graduates received the highest sample average but ANOVA alone does not identify which specific groups differ from each other. Further post-hoc tests, such as Tukey's HSD, would be needed to determine the pairwise differences.


````python
# post-hoc analysis 
from statsmodels.stats.multicomp import pairwise_tukeyhsd

# Perform Tukey's HSD test
tukey = pairwise_tukeyhsd(endog=df.melt()['value'], groups=df.melt()['variable'], alpha=0.05)
print(tukey)

# The summary table shows which pairs of groups have significantly different means. 
````

````python
# indepedent two-sample t-test 

import scipy.stats as stats

t_statistic, p_value = stats.ttest_ind(df['Finance'], df['Management'])
print("T-statistic:", t_statistic)
print("P-value:", p_value)
print("Reject null hypothesis" if p_value < 0.05 else "Fail to reject null hypothesis")   # type: ignore

# which group has a higher average number of job offers  
# Hypothesis0: mean of Finance - mean of Management = 0.
# Hypothesis1: mean of Finance - mean of Management > 0.

# Perform one-tailed t-test for the hypothesis that Finance has a higher mean than Management
t_statistic, p_value = stats.ttest_ind(df['Finance'], df['Management'], alternative='greater')
print("One-tailed T-statistic:", t_statistic)
print("One-tailed P-value:", p_value)
print("Reject null hypothesis" if p_value < 0.05 else "Fail to reject null hypothesis")  #type: ignore

higher_avg_group = 'Finance' if df['Finance'].mean() > df['Management'].mean() else 'Management' 

print("Group with higher average number of job offers:", higher_avg_group)
````

### Exerise 3

````python
# Randomized Block ANOVA: measurement error
import pandas as pd
from statsmodels.formula.api import ols
from statsmodels.stats.anova import anova_lm

measurements = pd.DataFrame({
    'Student': [1, 2, 3, 4],
    'Professor': [1.4, 3.1, 2.8, 3.4],
    'Male Student': [1.5, 2.6, 2.1, 3.6],
    'Female Student': [1.3, 2.4, 1.5, 2.9]
})

long_data = measurements.melt(
    id_vars='Student',
    var_name='Subject',
    value_name='Measurement_Error'
)

# Subject = treatment; Student = randomized block
model = ols(
    'Measurement_Error ~ C(Subject) + C(Student)',
    data=long_data
).fit()
anova_table = anova_lm(model, typ=2)

print(anova_table)
print(f"Treatment F-statistic: {anova_table.loc['C(Subject)', 'F']:.3f}")
print(f"Block F-statistic: {anova_table.loc['C(Student)', 'F']:.3f}")
print("Treatment decision:", "Reject H0" if anova_table.loc['C(Subject)', 'PR(>F)'] < 0.05 else "Do not reject H0") # type: ignore
print("Block decision:", "Reject H0" if anova_table.loc['C(Student)', 'PR(>F)'] < 0.05 else "Do not reject H0") # type: ignore
````

### Randomized Block ANOVA: Measurement Error

**a. Treatments and blocks**

- **Treatments:** The three subjects being measured: the professor, the male student, and the female student.
- **Blocks:** The four students who performed the measurements. Each student measured all three subjects, so student-to-student differences are controlled as blocks.

**b. Hypotheses**

Let $\mu_P$, $\mu_M$, and $\mu_F$ represent the mean measurement errors for the professor, male student, and female student.

- $H_0: \mu_P = \mu_M = \mu_F$
- $H_a:$ At least one mean measurement error differs.

**c. F-statistic**

From the ANOVA table,

$$
F = \frac{MS_{Treatments}}{MS_{Error}}
  = \frac{0.436}{0.086}
  \approx 5.078.
$$

**d. Decision at the 5% significance level**

The treatment p-value is $0.051$, which is greater than $\alpha=0.05$. Therefore, **do not reject $H_0$**. The block effect is significant because its p-value is $0.001$, but this does not change the treatment decision.

**e. Conclusion**

At the 5% significance level, there is **insufficient evidence to conclude that the mean measurement errors differ among the professor, male student, and female student**. The observed differences are not statistically significant at this level.
