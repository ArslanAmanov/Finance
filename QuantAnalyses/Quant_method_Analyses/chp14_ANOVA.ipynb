### Chapter 14: ANOVA (Analysis of Variance)

```python
#Two way ANOVA 

import pandas as pd
from statsmodels.formula.api import ols
from statsmodels.stats.anova import anova_lm
# Example data for two-way ANOVA

df= pd.DataFrame({
    'score': [
        70,72,75,78,80,82,
        65,68,70,71,73,75],
    'method': [
        'Traditional','Traditional','Online',
        'Online','Blended','Blended',
        'Traditional','Traditional','Online',
        'Online','Blended','Blended'],
    'class_size': [
        'Small','Small','Small',
        'Small','Small','Small',
        'Large','Large','Large',
        'Large','Large','Large'
    ]
})
```

```python
import statsmodels.api as sm
from statsmodels.formula.api import ols

model = ols(
    'score ~ C(method) * C(class_size)',
    data=df
).fit()

anova_table = sm.stats.anova_lm(model, typ=2)

print(anova_table)
```

Question 1 — One-way ANOVA

A company tests three sales training programs.


```python
# Determine:
# Whether the average sales performance differs between training programs.
# The ANOVA F-statistic.
# The p-value.
# If significant, perform Tukey HSD.
# Identify which training programs differ.


import pandas as pd
import statsmodels.api as sm
from statsmodels.formula.api import ols
from statsmodels.stats.multicomp import pairwise_tukeyhsd


df = pd.DataFrame({
    'score': [52, 55, 53, 54, 50, 56, 51, 60, 62, 61, 63, 59, 64, 62, 55, 57, 56, 58, 54, 59, 57],
    'program': ['A']*7 + ['B']*7 + ['C']*7
})

# Perform one-way ANOVA
model = ols('score ~ C(program)', data=df).fit()
anova_table = sm.stats.anova_lm(model, typ=2)
print(anova_table)

# If significant, perform Tukey HSD
tukey = pairwise_tukeyhsd(endog=df['score'], groups=df['program'], alpha=0.05)
print(tukey)

# Identify which training programs differ
print(tukey.summary())
```
