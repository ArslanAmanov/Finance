# Types of Data and Information

The objective of statistics is to extract information from data.

A variable is some characteristic of a population or sample.
For example, the mark on a statistics exam is a characteristic of statistics exams that is certainly of interest to 
readers of this book. Not all students achieve the same mark. The marks will vary from 
student to student, thus the name variable. The price of a stock is another variable. The 
prices of most stocks vary daily. We usually represent the name of the variable using 
uppercase letters such as X, Y, and Z.

The values of the variable are the possible observations of the variable. The 
values of statistics exam marks are the integers between 0 and 100 (assuming the exam 
is marked out of 100). The values of a stock price are real numbers that are usually 
measured in dollars and cents (sometimes in fractions of a cent). The values range from 
0 to hundreds of dollars.


These are the data from which we will extract the information we seek.
**Incidentally, data is plural for datum.** 
The mark of one student is a datum. 

When most people think of data, they think of sets of numbers. 
However, there are 
**three types of data: interval, nominal, and ordinal.**

# Interval Data: 
Interval data are real numbers, such as heights, weights, incomes, and distances. 

We also refer to this type of data as quantitative or numerical.

# Nominal Data:
Nominal data are names or labels. For example, the names of students in a class, 
the names of stocks in a stock market index, and the names of countries in the world are all examples of nominal data. 

The values of this variable are single, married, divorced, and widowed. 
Notice that the values are not numbers but instead are words that describe the categories.

**Nominal data are also called qualitative or categorical.**

# Ordinal Data:
Ordinal data are categorical data with a clear ordering of the categories. For example, the education levels of individuals (high school, bachelor's, master's, and doctorate) are ordinal data.

**Ordinal data are also called qualitative or categorical.**

# Range 
**Range** is a measure of dispersion that indicates the difference between the maximum and minimum values in a dataset. It provides a simple way to understand the spread of the data.

- Advantage of the range is that it is easy to calculate and understand. It gives a quick sense of the variability in the data.
- Disadvantage of the range is that it is sensitive to outliers. A single extreme value can significantly affect the range, making it less representative of the overall data distribution. 

- **Range = Largest value - Smallest value**

# Interpreting Standard Deviation Using the Empirical Rule

The **empirical rule**, also called the **68-95-99.7 rule**, applies to data that are approximately bell-shaped and symmetric (normally distributed).

If the mean is $\mu$ and the standard deviation is $\sigma$:

- About **68%** of observations lie within $1$ standard deviation of the mean: $\mu \pm \sigma$.
- About **95%** of observations lie within $2$ standard deviations of the mean: $\mu \pm 2\sigma$.
- About **99.7%** of observations lie within $3$ standard deviations of the mean: $\mu \pm 3\sigma$.

## Quick Example

Suppose exam scores have a mean of $70$ and a standard deviation of $5$:

- About 68% of scores lie between $65$ and $75$: $70 \pm 5$.
- About 95% of scores lie between $60$ and $80$: $70 \pm 10$.
- About 99.7% of scores lie between $55$ and $85$: $70 \pm 15$.

## Interpretation

A smaller standard deviation means that observations are concentrated close to the mean. A larger standard deviation means that observations are more widely spread out.

The empirical rule should be used only when the distribution is approximately normal. For strongly skewed or irregular data, use other methods such as percentiles or the interquartile range.


# Chebyshev's Theorem

**Chebyshev's theorem** gives a minimum proportion of observations that must lie within a specified number of standard deviations from the mean. It applies to **any distribution**, whether the data are symmetric, skewed, or irregular.

For any value of $k > 1$, at least the following proportion of observations lies within $k$ standard deviations of the mean:

$$
1 - \frac{1}{k^2}
$$

The interval is:

$$
\mu - k\sigma \leq X \leq \mu + k\sigma
$$

where $\mu$ is the mean and $\sigma$ is the standard deviation.

## Common Values

| Number of standard deviations | Minimum proportion | Minimum percentage |
|---:|---:|---:|
| $k = 2$ | $1 - \frac{1}{2^2} = 0.75$ | **75%** |
| $k = 3$ | $1 - \frac{1}{3^2} \approx 0.8889$ | **88.89%** |
| $k = 4$ | $1 - \frac{1}{4^2} = 0.9375$ | **93.75%** |

## Example

Suppose a dataset has a mean of $70$ and a standard deviation of $5$. Chebyshev's theorem guarantees that at least what percentage of observations lie between $60$ and $80$?

Here, $k = 2$ because:

$$
70 - 2(5) = 60
$$

$$
70 + 2(5) = 80
$$

The minimum proportion is:

$$
1 - \frac{1}{2^2} = 1 - \frac{1}{4} = 0.75
$$

Therefore, at least **75% of the observations** lie between $60$ and $80$.

## Chebyshev's Theorem vs. the Empirical Rule

- **Chebyshev's theorem:** applies to every distribution, but gives only a minimum guarantee.
- **Empirical rule:** applies to approximately normal distributions and gives the typical percentages of 68%, 95%, and 99.7%.

For example, within $2$ standard deviations, Chebyshev's theorem guarantees at least 75%, while the empirical rule estimates approximately 95% for a normal distribution.


# Coefficient of Variation

The **coefficient of variation (CV)** measures relative variability. It compares the standard deviation with the mean and is useful when datasets have different units or different means.

For a population:

$$
CV = \frac{\sigma}{\mu} \times 100\%
$$

For a sample:

$$
CV = \frac{s}{\bar{x}} \times 100\%
$$

where $\sigma$ is the population standard deviation, $\mu$ is the population mean, $s$ is the sample standard deviation, and $\bar{x}$ is the sample mean.

## Example

Suppose a dataset has a mean of $50$ and a standard deviation of $5$:

$$
CV = \frac{5}{50} \times 100\% = \boxed{10\%}
$$

This means that the standard deviation is equal to 10% of the mean.

## Comparing Relative Variability

- Dataset A: mean $100$, standard deviation $10$

  $$
  CV_A = \frac{10}{100} \times 100\% = 10\%
  $$

- Dataset B: mean $20$, standard deviation $4$

  $$
  CV_B = \frac{4}{20} \times 100\% = 20\%
  $$

Although Dataset A has the larger standard deviation, Dataset B has greater relative variability because its CV is higher.

## Common Uses

- Comparing risk between investments with different average returns.
- Comparing variation in measurements with different units or scales.
- Comparing consistency of machines, processes, or production results.

A lower CV indicates greater consistency relative to the mean. The CV is generally used for positive, ratio-scale data with a meaningful zero. It is not appropriate when the mean is zero or close to zero.

# Harmonic Mean

The **harmonic mean** is an average used mainly for **rates, ratios, and per-unit quantities**. It is especially useful when the same amount of work, distance, or quantity is involved.

For $n$ positive values $x_1, x_2, \ldots, x_n$:

$$
H = \frac{n}{\frac{1}{x_1}+\frac{1}{x_2}+\cdots+\frac{1}{x_n}}
$$

For two values $a$ and $b$:

$$
H = \frac{2ab}{a+b}
$$

## Example: Average Speed

A car travels equal distances at $60$ km/h and $40$ km/h. The average speed is the harmonic mean:

$$
H = \frac{2(60)(40)}{60+40}
= \frac{4800}{100}
= 48\text{ km/h}
$$

The average speed is **48 km/h**, not 50 km/h, because the car spends more time traveling at the slower speed.

## Where Is It Used?

- **Average speed:** combining speeds over equal distances.
- **Average price per unit:** comparing equal quantities purchased at different unit prices.
- **Finance:** averaging ratios such as price-to-earnings ratios or return multiples.
- **Productivity:** combining worker or machine rates, such as tasks per hour.
- **Computer science and networking:** averaging data-transfer rates, processing speeds, or throughput.
- **Machine learning:** calculating the F1 score from precision and recall.

The F1 score is the harmonic mean of precision and recall:

$$
F_1 = \frac{2(\text{precision})(\text{recall})}
{\text{precision}+\text{recall}}
$$

The harmonic mean gives more influence to smaller values. Therefore, a high value for one measure cannot completely compensate for a very low value for another measure.

## Sample Exercises

1. Find the harmonic mean of $4$ and $12$.
2. Find the harmonic mean of $3$, $6$, and $12$.
3. A cyclist travels equal distances at $20$ km/h and $30$ km/h. Find the average speed.
4. A machine completes tasks at rates of $10$, $15$, and $30$ tasks per hour. Find the harmonic mean of the rates.
5. Compare the arithmetic mean and harmonic mean of $5$ and $20$.
6. A classification model has a precision of $0.80$ and a recall of $0.50$. Calculate its F1 score.

## Answers

1. $H = 6$
2. $H = 6$
3. $H = 24$ km/h
4. $H = 15$ tasks per hour
5. Arithmetic mean: $12.5$; harmonic mean: approximately $8.0$
6. $F_1 = \frac{2(0.80)(0.50)}{0.80+0.50} \approx 0.615$


## Common Application Examples

### 1. Average Speed

Use the harmonic mean when equal distances are traveled at different speeds. For example, if a driver travels one half of a journey at $60$ km/h and the other half at $40$ km/h, the correct average speed is:

$$
H = \frac{2(60)(40)}{60+40} = 48\text{ km/h}
$$

### 2. Average Unit Price

The harmonic mean can be used when equal quantities are purchased at different prices per unit. For example, if the same quantity is bought at $10$ dollars per kilogram and $15$ dollars per kilogram:

$$
H = \frac{2(10)(15)}{10+15} = 12\text{ dollars per kilogram}
$$

### 3. Worker or Machine Productivity

If workers or machines complete equal amounts of work at different rates, the harmonic mean gives the combined average rate. For rates of $10$ and $20$ tasks per hour:

$$
H = \frac{2(10)(20)}{10+20} \approx 13.33\text{ tasks per hour}
$$

### 4. Finance and Investment Ratios

The harmonic mean is useful for averaging ratios, such as price-to-earnings ratios, when each investment or company should receive equal weight. It prevents a very large ratio from dominating the average.

### 5. Networking and Computer Performance

The harmonic mean can summarize data-transfer rates, processing speeds, or system throughput. A slow connection or component has a strong effect on the overall average, which reflects real performance limitations.

### 6. Machine Learning Classification

The F1 score uses the harmonic mean of precision and recall. It is useful when both false positives and false negatives matter:

$$
F_1 = \frac{2PR}{P+R}
$$

where $P$ is precision and $R$ is recall. If precision is $0.80$ and recall is $0.50$:

$$
F_1 = \frac{2(0.80)(0.50)}{0.80+0.50} \approx 0.615
$$

### Important Rule

Use the harmonic mean for rates or ratios when the quantities being compared have equal weights, equal distances, equal amounts of work, or equal quantities. Do not use it automatically for ordinary measurements such as heights, ages, or exam scores.


# Geometric Mean

The **geometric mean** is an average used for positive values that are multiplied together. It is especially useful for **growth rates, investment returns, percentages, and other multiplicative changes**.

For $n$ positive values $x_1, x_2, \ldots, x_n$:

$$
G = \sqrt[n]{x_1x_2\cdots x_n}
$$

For two values:

$$
G = \sqrt{ab}
$$

### Example

The geometric mean of $4$ and $16$ is:

$$
G = \sqrt{4 \times 16} = \sqrt{64} = 8
$$

In finance, the geometric mean is useful for finding an average investment growth rate over multiple periods because returns compound. For example, growth factors of $1.10$ and $1.20$ produce an average growth factor of:

$$
G = \sqrt{1.10 \times 1.20} \approx 1.149
$$

This represents an average compound growth rate of approximately $14.9\%$ per period.

### Important Note

Use the geometric mean for positive values and compounded changes. Unlike the arithmetic mean, it reflects multiplication and compounding rather than simple addition.

```python
# compute the geometric mean of the following rates of return 
# a) .25, -.10, .50, .40 
# b) .50, .30, -.50, -.25
# c) .30, .50, .10, .20 

import numpy as np 
import statistics as st
a_returns = [0.25, -0.10, 0.50, 0.40]
b_returns = [0.50, 0.30, -0.50, -0.25]
c_returns = [0.30, 0.50, 0.10, 0.20]

def geometric_mean(returns):
    # Convert the returns to a numpy array
    returns_array = np.array(returns)
    
    # Calculate the product of (1 + return) for each return
    product = np.prod(1 + returns_array)
    
    # Calculate the geometric mean
    n = len(returns_array)
    geo_mean = product**(1/n) - 1
    
    return geo_mean

print("Geometric Mean for a):", geometric_mean(a_returns))
print("Geometric Mean for b):", geometric_mean(b_returns))
print("Geometric Mean for c):", geometric_mean(c_returns))
```

An investment of $1,000 produces annual rates of return of $0.10$, $0.50$, and $-0.30$ over three years.

## a. Value of the Investment After Each Year

$$
V_1 = 1000(1+0.10) = 1000(1.10) = \boxed{\$1,100}
$$

### Year 2

$$
V_2 = 1100(1+0.50) = 1100(1.50) = \boxed{\$1,650}
$$

### Year 3

$$
V_3 = 1650(1-0.30) = 1650(0.70) = \boxed{\$1,155}
$$

The final value is $\boxed{\$1,155}$.

## b. Geometric Mean and Arithmetic Mean

### Geometric Mean

The geometric mean return accounts for compounding:

$$
G = (1.10 \times 1.50 \times 0.70)^{1/3} - 1
$$

$$
G = 1.0492 - 1 \approx \boxed{4.92\%}
$$

### Arithmetic Mean

$$
A = \frac{0.10+0.50-0.30}{3}
= \frac{0.30}{3}
= \boxed{10\%}
$$

## c. Investment Values Using the Geometric Mean

If each year's return equals the geometric mean, each year grows by approximately $4.92\%$:

- After Year 1: $1000(1.0492) \approx \boxed{\$1,049.21}$
- After Year 2: $1000(1.0492)^2 \approx \boxed{\$1,100.83}$
- After Year 3: $1000(1.0492)^3 = \boxed{\$1,155.00}$

If each year's return equals the arithmetic mean of $10\%$:

- After Year 1: $1000(1.10) = \boxed{\$1,100}$
- After Year 2: $1000(1.10)^2 = \boxed{\$1,210}$
- After Year 3: $1000(1.10)^3 = \boxed{\$1,331}$

## e. Why the Geometric Mean Is Better

The geometric mean is a better measure of average investment return because investment returns compound over time. It accounts for both gains and losses and gives the constant annual return that produces the same final investment value.

The arithmetic mean ignores compounding. In this example, it suggests a $10\%$ average return, but applying $10\%$ each year produces $1,331$, which is greater than the actual final value of $1,155$. The geometric mean produces the correct equivalent annual return of approximately $4.92\%$.

# Exercise 4.33: Standard Deviation versus MAD

If a sample has very large deviations, the **sample standard deviation** will generally be larger than the **mean absolute deviation (MAD)**.

The reason is that standard deviation uses **squared deviations**:

$$
 s = \sqrt{\frac{\sum (x_i-\bar{x})^2}{n-1}}
$$

MAD uses absolute deviations:

$$
MAD = \frac{\sum \lvert x_i-\bar{x}\rvert}{n}
$$

Squaring makes large deviations grow much faster. For example, a deviation of $10$ contributes $100$ to the sum of squared deviations but only $10$ to the sum of absolute deviations.

Therefore, the sample standard deviation is more strongly affected by extreme values, while MAD is less sensitive to outliers. Both measures use the same units as the original data, but the standard deviation gives greater weight to unusually large deviations.

\bar{x} = \frac{6+23+31+14+11}{5}
= \frac{85}{5}
= 17
$$

## Step 2: Calculate Deviations and Absolute Deviations

| Value $x_i$ | Deviation $x_i-\bar{x}$ | Squared deviation | Absolute deviation $\lvert x_i-\bar{x}\rvert$ |
|---:|---:|---:|---:|
| 6 | -11 | 121 | 11 |
| 23 | 6 | 36 | 6 |
| 31 | 14 | 196 | 14 |
| 14 | -3 | 9 | 3 |
| 11 | -6 | 36 | 6 |
| **Total** | **0** | **398** | **40** |

## Sample Standard Deviation

Because these are sample data, divide the squared-deviation total by $n-1$:

$$
 s^2 = \frac{398}{5-1} = \frac{398}{4} = 99.5
$$

$$
 s = \sqrt{99.5} \approx \boxed{9.97}
$$

## Mean Absolute Deviation

$$
MAD = \frac{\sum \lvert x_i-\bar{x}\rvert}{n}
= \frac{40}{5}
= \boxed{8}
$$

Therefore:

- Sample mean: $\boxed{17}$
- Sample standard deviation: $\boxed{9.97}$
- Mean absolute deviation (MAD): $\boxed{8}$


```python
import numpy as np

data = np.array([6, 23, 31, 14, 11]) 

mean = np.mean(data)
variance = np.var(data, ddof=1)
standard_deviation = np.sqrt(variance)
mad = np.mean(np.abs(data - mean))

print("Mean:", mean)
print("Variance:", variance)
print("Standard Deviation:", standard_deviation)
print("Mean Absolute Deviation (MAD):", mad)
```

# Exercise 4.39: One-Sided Proportions Using the Empirical Rule

Exercise 4.38 gives a bell-shaped distribution with:

- Mean: $\mu=50$
- Standard deviation: $\sigma=4$

For a bell-shaped distribution, the empirical rule tells us:

- About 68% of observations lie between $\mu-\sigma$ and $\mu+\sigma$.
- About 95% lie between $\mu-2\sigma$ and $\mu+2\sigma$.

Because the distribution is symmetric, the observations outside each central interval are split equally between the two tails.

## a. Proportion Less Than 46

First, determine how many standard deviations 46 is below the mean:

$$
50-46=4=1\sigma
$$

Therefore, 46 is one standard deviation below the mean. About 68% lie between 46 and 54, so 32% lie outside this interval:

$$
100\%-68\%=32\%
$$

The lower and upper tails are equal, so:

$$
\frac{32\%}{2}=\boxed{16\%}
$$

Approximately **16% of observations are less than 46**.

## b. Proportion Less Than 58

Determine how many standard deviations 58 is above the mean:

$$
58-50=8=2\sigma
$$

About 95% of observations lie between 42 and 58. Therefore, 5% lie outside this interval, with 2.5% in the upper tail:

$$
\frac{100\%-95\%}{2}=\frac{5\%}{2}=2.5\%
$$

The proportion less than 58 is:

$$
100\%-2.5\%=\boxed{97.5\%}
$$

Approximately **97.5% of observations are less than 58**.

## c. Proportion Greater Than 54

Determine how many standard deviations 54 is above the mean:

$$
54-50=4=1\sigma
$$

About 68% lie between 46 and 54, leaving 32% outside the interval. Half is in the upper tail:

$$
\frac{32\%}{2}=\boxed{16\%}
$$

Approximately **16% of observations are greater than 54**.

## Summary

| Question | Standardized location | Approximate proportion |
|---|---:|---:|
| Less than 46 | $\mu-1\sigma$ | **16%** |
| Less than 58 | $\mu+2\sigma$ | **97.5%** |
| Greater than 54 | $\mu+1\sigma$ | **16%** |


# Exercise 4.40: Applying Chebyshev's Theorem

The histogram is extremely skewed, so the empirical rule cannot be used. Instead, apply Chebyshev's theorem:

$$
\text{Minimum proportion} = 1-\frac{1}{k^2}
$$

where $k$ is the number of standard deviations from the mean. Here, the mean is $\mu=70$ and the standard deviation is $\sigma=12$.

## a. Observations Between 46 and 94

The distance from the mean to either endpoint is:

$$
70-46=24, \qquad 94-70=24
$$

Since:

$$
\frac{24}{12}=2
$$

the interval is $\mu\pm2\sigma$. Using Chebyshev's theorem:

$$
1-\frac{1}{2^2}=1-\frac{1}{4}=0.75
$$

Therefore, the minimum proportion of observations between 46 and 94 is:

$$
\boxed{75\%}
$$

## b. Observations Between 34 and 106

The distance from the mean to either endpoint is:

$$
70-34=36, \qquad 106-70=36
$$

Since:

$$
\frac{36}{12}=3
$$

the interval is $\mu\pm3\sigma$. Using Chebyshev's theorem:

$$
1-\frac{1}{3^2}=1-\frac{1}{9}=\frac{8}{9}\approx0.8889
$$

Therefore, the minimum proportion of observations between 34 and 106 is:

$$
\boxed{88.89\%}
$$

## Summary

| Interval | Number of standard deviations | Minimum proportion |
|---|---:|---:|
| 46 to 94 | $\pm2\sigma$ | **75%** |
| 34 to 106 | $\pm3\sigma$ | **88.89%** |
